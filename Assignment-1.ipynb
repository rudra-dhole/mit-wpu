```python
students = {
    101: ("Alice", "Computer Science", [85, 90, 88]),
    102: ("Bob", "Electrical", [78, 82, 80]),
    103: ("Charlie", "Mechanical", [92, 89, 94])
}

def display_students(records, title="Student Records"):
    """Helper function to print all student records clearly."""
    print(f"\n--- {title} ---")
    print(f"{'Roll No':<10} | {'Name':<12} | {'Branch':<18} | {'Marks'}")
    for roll_no, details in records.items():
        name, branch, marks = details
        print(f"{roll_no:<10} | {name:<12} | {branch:<18} | {marks}")

display_students(students, "Initial Student Records")

new_roll_no = 104
students[new_roll_no] = ("Diana", "Information Tech", [88, 91, 85])
print(f"\n[+] Added record for Roll Number {new_roll_no}.")

del_roll_no = 102
if del_roll_no in students:
    deleted_student = students.pop(del_roll_no)
    print(f"[-] Deleted record for Roll Number {del_roll_no} ({deleted_student[0]}).")

update_roll_no = 101
if update_roll_no in students:
    current_name, _, _ = students[update_roll_no]
    updated_branch = "AI & Data Science"
    updated_marks = [90, 94, 92]
    
    students[update_roll_no] = (current_name, updated_branch, updated_marks)
    print(f"[*] Updated details for Roll Number {update_roll_no}.")

display_students(students, "Final Student Records")
```
