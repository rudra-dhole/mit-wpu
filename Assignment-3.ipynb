```python
def is_right_angled(a, b, c):
    sides = sorted([a, b, c])
    if sides[0] + sides[1] <= sides[2]:
        return "Invalid Triangle: Sum of two smaller sides must be greater than the third."
    if round(sides[0]**2 + sides[1]**2, 5) == round(sides[2]**2, 5):
        return True
    else:
        return False


def main():
    print("--- Right-Angled Triangle Checker ---")
    try:
        side1 = float(input("Enter length of first side: "))
        side2 = float(input("Enter length of second side: "))
        side3 = float(input("Enter length of third side: "))

        if side1 <= 0 or side2 <= 0 or side3 <= 0:
            print("Side lengths must be positive numbers.")
            return

        result = is_right_angled(side1, side2, side3)

        if result is True:
            print(f"\nThe sides ({side1}, {side2}, {side3}) form a Right-Angled Triangle.")
        elif result is False:
            print(f"\nThe sides ({side1}, {side2}, {side3}) do NOT form a Right-Angled Triangle.")
        else:
            print(f"\n{result}")

    except ValueError:
        print("Invalid input! Please enter numerical values for side lengths.")


if __name__ == "__main__":
    main()
```
