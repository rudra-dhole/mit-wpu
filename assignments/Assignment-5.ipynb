```python
def validate_pan(pan_number):
    # 1. Check total length
    if len(pan_number) != 10:
        return "Invalid PAN Number"
    
    # 2. Check first 5 characters (A-Z)
    first_five = pan_number[:5]
    if not (first_five.isalpha() and first_five.isupper()):
        return "Invalid PAN Number"
    
    # 3. Check next 4 characters (0-9)
    next_four = pan_number[5:9]
    if not next_four.isdigit():
        return "Invalid PAN Number"
    
    # 4. Check the last character (A-Z)
    last_char = pan_number[9]
    if not (last_char.isalpha() and last_char.isupper()):
        return "Invalid PAN Number"
    
    return "Valid PAN Number"

def main():
    user_input = input("Enter PAN Number: ").strip()
    result = validate_pan(user_input)
    print(result)


if __name__ == "__main__":
    main()
```
